```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation

from IPython.display import HTML
```

```python
fig, ax = plt.subplots()
ax.set_xlim(-20, 20)
ax.set_ylim(-20, 20)
ax.set_aspect('equal')
ax.axis('off')
t = np.linspace(0, 2 * np.pi, 1000)
x = 16 * np.sin(t)**3
y = 13 * np.cos(t) - 5 * np.cos(2*t) - 2 * np.cos(3*t) - np.cos(4*t)

heart, = ax.plot([], [], 'r')

def init():
    heart.set_data([], [])
    return heart,
def update(frame):
    heart.set_data(x[:frame], y[:frame])
    return heart,

ani = animation.FuncAnimation(fig, update, frames=len(t), init_func=init, blit=True, interval=5)

#ani.save('heart_animation.mp4', writer='ffmpeg')


HTML(ani.to_jshtml(default_mode="loop"))
```
