```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

## Case for $\varphi = 1$.

Distribution uniforme de $\Delta _p$ entre $\Delta _p^{\text{min}}$ et $\Delta _p^{\text{max}}$

```python
alpha = 2.
beta = 1.
gamma = 4.
t_star = 0.

f = 3.
E = 1.

delta = beta * gamma / (beta + gamma)

N = 10000.
s = 4000.
M = 3.

Delta_d = 1
Delta_p_min, Delta_p_max = 5, 10

F_generic = sp.stats.uniform(loc=Delta_p_min, scale=Delta_p_max - Delta_p_min)
Delta_p_full = F_generic.ppf(M / (M+1))

def equilibrium_modal_split(rho_d, rho_p):
    if Delta_p_full < f + Delta_d + rho_d - rho_p:
        raise ValueError("No equilibrium")
    elif Delta_p_min > f + Delta_d + rho_d - rho_p:
        return 0., N, Delta_p_min
    else:
        N_p = N * F_generic.cdf(f + Delta_d + rho_d - rho_p)
        N_d = N - N_p
        Delta_p_switch = F_generic.cdf(N_p / N)
        return N_p, N_d, Delta_p_switch
```

```python
rho_d = 0
rho_p = 0
    
fig, (ax_scale, ax) = plt.subplots(2,1)

ax_scale.axis("off")
pos_1 = ax_scale.get_position()
ax_scale.set_position([pos_1.x0, pos_1.y0 + pos_1.height * 0.1, pos_1.width, pos_1.height * 0.2])

ax_scale.plot([Delta_p_min, Delta_p_min], [-1, 1], color="black")
ax_scale.plot([Delta_p_max, Delta_p_max], [-1, 1], color="black")
ax_scale.plot([Delta_p_full, Delta_p_full], [-1, 1], color="black")
ax_scale.plot([Delta_p_min, Delta_p_max], [0, 0], color="black")

ax_scale.set_title(r"Position of $f + \overline{\Delta_d} + \rho_d - \rho_p$ relative to $\Delta_p$ distribution")
ax_scale.text(Delta_p_min, -1.5, r"$\Delta_p^{\text{min}}$")
ax_scale.text(Delta_p_full, -1.5, r"$\Delta_p^{\text{full}}$")
ax_scale.text(Delta_p_max, -1.5, r"$\Delta_p^{\text{max}}$")

ax_scale.scatter([f + Delta_d + rho_d - rho_p], [0], color="red")


N_p, N_d, Delta_p_switch = equilibrium_modal_split(rho_d, rho_p)
ax.set_title(r"Equilibrium modal split ($N_p / N = {N_p:.0f}\% $, $N_d / N = {N_d:.0f}\% $)".format(N_p=N_p/N * 100, N_d = N_d/N * 100))

t_d_0 = t_star - (N_d / s) * gamma / (beta + gamma)
t_d_tilde = t_star - (N_d / s) * (beta*gamma) / (alpha * (beta + gamma))
t_d_1 = t_star + (N_d / s) * beta / (beta + gamma)

pos_2 = ax.get_position()
ax.set_position([pos_2.x0, pos_2.y0 - pos_2.height, pos_1.width, pos_1.height * 2])


xticks = {r"$t_d^0$": t_d_0,
         r"$\tilde{t}_d$": t_d_tilde,
         r"$t^*$": t_star,
         r"$t_d^1$": t_d_1
        }

xticklabels = list(xticks.keys())
xtickvalues = list(xticks.values())

yticklabels = ["0", "N"]
ytickvalues = [0, N]

ax.set_xticks(xtickvalues)
ax.set_xticklabels(xticklabels)
ax.set_yticks(ytickvalues)
ax.set_yticklabels(yticklabels)

ax.set_ylim(bottom=0 - N/10, top=N*1.1)
ax.set_xlim(left=t_d_0 - (t_d_1 - t_d_0)/10, right=t_d_1 + (t_d_1 - t_d_0)/10)

x = np.linspace(t_d_0, t_d_1, 1000)
x1 = np.linspace(t_d_0, t_d_tilde, 1000)
x2 = np.linspace(t_d_tilde, t_d_1, 1000)

ax.plot(x, s*(x-t_d_0), color = "black", linestyle="--", label=r"Arrivals (slope = $s$)")
ax.plot(x1, (s + beta * s / (alpha - beta)) * (x1-t_d_0), color="red", label=r"Departures before $\tilde{t}_d$ (slope = $s + \frac{\beta s}{\alpha - \beta}$)")
ax.plot(x2, (s - gamma * s / (alpha + gamma)) * (x2-t_d_tilde) + (t_d_tilde - t_d_0) * (s + beta * s / (alpha - beta)), color = "green", label=r"Departures after $\tilde{t}_d$ (slope = $s - \frac{\gamma s}{\alpha + \gamma}$")

ax.legend(loc="lower right")

ax.set_xlabel("Time (t)")
ax.set_ylabel("Cumulative departures and arrivals")
```

## Case $0<\varphi<1$

```python
alpha = 2.
beta = 1.
gamma = 4.
t_star = 0.

f = 3.
E = 1.

delta = beta * gamma / (beta + gamma)

N = 10000.
s = 4000.
M = 3.
varphi = 0.3

Delta_d = 1
Delta_p_min, Delta_p_max = 5, 10

F_generic = sp.stats.uniform(loc=Delta_p_min, scale=Delta_p_max - Delta_p_min)

Delta_p_full = F_generic.ppf(M / (M+1))
```

```python
fig, ax = plt.subplots()

#ax.axis("off")
ax.set_xticks([])
ax.set_yticks([])
ax.set_xlabel(r"$\rho_d$")
ax.set_ylabel(r"$\rho_p$")
ax.set_title("Equilibrium phase diagram")

rho_d_min = -delta*N/(varphi * s) - Delta_d
rho_d_max = delta*N/((1-varphi)*s) - Delta_d
rho_d_min_frame = rho_d_min - (rho_d_max - rho_d_min) / 3
rho_d_max_frame = rho_d_max + (rho_d_max - rho_d_min) / 3

rho_p_min = rho_d_min_frame + f + Delta_d - Delta_p_full
rho_p_max = rho_d_max + f + Delta_d - Delta_p_min
rho_p_min_frame = rho_p_min - (rho_p_max - rho_p_min) / 3
rho_p_max_frame = rho_p_max + (rho_p_max - rho_p_min) / 3

ax.set_xlim(left=rho_d_min_frame, right=rho_d_max_frame)
ax.set_ylim(bottom=rho_p_min_frame, top=rho_p_max_frame)

# Plot only solo line
ax.plot([rho_d_max, rho_d_max],
        [rho_p_min_frame, rho_p_max_frame],
        color="black")

# Plot no passenger line
rho_d_no_pass = np.linspace(rho_d_min_frame, rho_d_max, 1000)
rho_p_no_pass = rho_d_no_pass + f + Delta_d - Delta_p_min
ax.plot(rho_d_no_pass, rho_p_no_pass, color="black")

# Plot no solo no passenger line
ax.plot([rho_d_min, rho_d_min],
        [rho_d_min + f + Delta_d - Delta_p_min, rho_p_max_frame],
        color="black")


# Plot no solo with passenger line
N_d_no_solo_with_passenger = N * np.linspace(1/(M+1), 1, 1000)
rho_d_no_solo_with_passenger = -delta * N_d_no_solo_with_passenger / (varphi * s) - Delta_d
rho_p_no_solo_with_passenger = rho_d_no_solo_with_passenger + f + Delta_d - F_generic.ppf((N-N_d_no_solo_with_passenger) / N)
ax.plot(rho_d_no_solo_with_passenger, rho_p_no_solo_with_passenger, color="black")

# Plot no solo infeasible line
rho_d_no_solo_infeasible = np.linspace(rho_d_min_frame, -(delta / (varphi * s)) * (N/(M+1)) - Delta_d, 1000)
rho_p_no_solo_infeasible = rho_d_no_solo_infeasible + f + Delta_d - Delta_p_full
ax.plot(rho_d_no_solo_infeasible, rho_p_no_solo_infeasible, color="black", linestyle="--")

# Plot solo and carpool infeasible line
rho_d_solo_carpool_infeasible = np.linspace(-(delta / (varphi * s)) * (N/(M+1)) - Delta_d, rho_d_max, 1000)

Delta_p_full_line = F_generic.ppf((1/N) * (M/(M+1)) * \
                                    (N - \
                                        (rho_d_solo_carpool_infeasible + (delta/(varphi*s)) * (N/(M+1)) + Delta_d ) / \
                                        ((delta / s) * (1/(1-varphi) + 1/(varphi * (M+1))))\
                                    )\
                                 )

rho_p_solo_carpool_infeasible = rho_d_solo_carpool_infeasible + f + Delta_d - Delta_p_full_line
ax.plot(rho_d_solo_carpool_infeasible, rho_p_solo_carpool_infeasible, color="black", linestyle="--")



# Annotate diagram

prop_x = rho_d_max_frame - rho_d_min_frame
prop_y = rho_p_max_frame - rho_p_min_frame

# Draw arrows
ax.arrow(rho_d_max, rho_p_min_frame + prop_y/2, prop_x/30, 0, head_width=prop_x/20, head_length=prop_y/50, fc='k', ec='k')
ax.arrow(rho_d_min, rho_p_min_frame + 2 * prop_y/3, -prop_x/30, 0, head_width=prop_x/20, head_length=prop_y/50, fc='k', ec='k')
ax.arrow(rho_d_min_frame + (rho_d_min - rho_d_min_frame) / 2, rho_d_min_frame + (rho_d_min - rho_d_min_frame) / 2 + f + Delta_d - Delta_p_min, 0,  prop_y/30, head_width=prop_y/60, head_length=prop_x/15, fc='k', ec='k')
ax.arrow(rho_d_min + (rho_d_max - rho_d_min) / 2, rho_d_min + (rho_d_max - rho_d_min) / 2 + f + Delta_d - Delta_p_min, 0,  prop_y/30, head_width=prop_y/60, head_length=prop_x/15, fc='k', ec='k')
ax.arrow(rho_d_min_frame + 0.25*prop_x, rho_p_min_frame + 0.46*prop_y, 0,  -prop_y/30, head_width=prop_y/60, head_length=prop_x/15, fc='k', ec='k')

# Name curves

ax.text(rho_d_min_frame + 0.81*prop_x,
        rho_p_min_frame + 0.95*prop_y,
        "(1)",
        color="black")

ax.text(rho_d_min_frame + 0.35*prop_x,
        rho_p_min_frame + 0.58*prop_y,
        "(2)",
        color="black")

ax.text(rho_d_min_frame + 0.02*prop_x,
        rho_p_min_frame + 0.39*prop_y,
        "(2)",
        color="black")

ax.text(rho_d_min_frame + 0.21*prop_x,
        rho_p_min_frame + 0.95*prop_y,
        "(3)",
        color="black")

ax.text(rho_d_min_frame + 0.45*prop_x,
        rho_p_min_frame + 0.51*prop_y,
        "(4)",
        color="black")

ax.text(rho_d_min_frame + 0.65*prop_x,
        rho_p_min_frame + 0.6*prop_y,
        "(5)",
        color="black")

ax.text(rho_d_min_frame + 0.2*prop_x,
        rho_p_min_frame + 0.28*prop_y,
        "(6)",
        color="black")


# Name spaces

ax.text(rho_d_min_frame + 0.89*prop_x,
        rho_p_min_frame + 0.65*prop_y,
        "A",
        fontsize=15,
        color="black")

ax.text(rho_d_min_frame + 0.5*prop_x,
        rho_p_min_frame + 0.8*prop_y,
        "B",
        fontsize=15,
        color="black")

ax.text(rho_d_min_frame + 0.09*prop_x,
        rho_p_min_frame + 0.75*prop_y,
        "C",
        fontsize=15,
        color="black")

ax.text(rho_d_min_frame + 0.09*prop_x,
        rho_p_min_frame + 0.31*prop_y,
        "D",
        fontsize=15,
        color="black")

ax.text(rho_d_min_frame + 0.5*prop_x,
        rho_p_min_frame + 0.55*prop_y,
        "E",
        fontsize=15,
        color="black")

ax.text(rho_d_min_frame + 0.6*prop_x,
        rho_p_min_frame + 0.25*prop_y,
        "F",
        fontsize=15,
        color="black")
```

```python
alpha = 2.
beta = 1.
gamma = 4.
t_star = 0.

f = 3.
E = 1.

delta = beta * gamma / (beta + gamma)

N = 10000.
s = 4000.
M = 3.
varphi = 0.3

Delta_d = 1
Delta_p_min, Delta_p_max = 5, 10

F_generic = sp.stats.trapezoid(c=0.4, d=0.6, loc=Delta_p_min, scale=Delta_p_max - Delta_p_min)

Delta_p_full = F_generic.ppf(M / (M+1))
```
