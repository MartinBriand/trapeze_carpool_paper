```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

## Case for $\varphi = 1$.

Distribution uniforme de $\Delta _p$ entre $\Delta _p^{\text{min}}$ et $\Delta _p^{\text{max}}$

```python
alpha = 2.
beta = 1.
gamma = 4.
t_star = 0.

f = 3.
E = 1.

delta = beta * gamma / (beta + gamma)

N = 10000.
s = 4000.
M = 3.

Delta_d = 1
Delta_p_min, Delta_p_max = 5, 10

F_generic = sp.stats.uniform(loc=Delta_p_min, scale=Delta_p_max - Delta_p_min)
Delta_p_full = F_generic.ppf(M / (M+1))

def equilibrium_modal_split(rho_d, rho_p):
    if Delta_p_full < f + Delta_d + rho_d - rho_p:
        raise ValueError("No equilibrium")
    elif Delta_p_min > f + Delta_d + rho_d - rho_p:
        return 0., N, Delta_p_min
    else:
        N_p = N * F_generic.cdf(f + Delta_d + rho_d - rho_p)
        N_d = N - N_p
        Delta_p_switch = F_generic.cdf(N_p / N)
        return N_p, N_d, Delta_p_switch
```

```python
rho_d = 0
rho_p = 0
    
fig, (ax_scale, ax) = plt.subplots(2,1)

ax_scale.axis("off")
pos_1 = ax_scale.get_position()
ax_scale.set_position([pos_1.x0, pos_1.y0 + pos_1.height * 0.1, pos_1.width, pos_1.height * 0.2])

ax_scale.plot([Delta_p_min, Delta_p_min], [-1, 1], color="black")
ax_scale.plot([Delta_p_max, Delta_p_max], [-1, 1], color="black")
ax_scale.plot([Delta_p_full, Delta_p_full], [-1, 1], color="black")
ax_scale.plot([Delta_p_min, Delta_p_max], [0, 0], color="black")

ax_scale.set_title(r"Position of $f + \overline{\Delta_d} + \rho_d - \rho_p$ relative to $\Delta_p$ distribution")
ax_scale.text(Delta_p_min, -1.5, r"$\Delta_p^{\text{min}}$")
ax_scale.text(Delta_p_full, -1.5, r"$\Delta_p^{\text{full}}$")
ax_scale.text(Delta_p_max, -1.5, r"$\Delta_p^{\text{max}}$")

ax_scale.scatter([f + Delta_d + rho_d - rho_p], [0], color="red")


N_p, N_d, Delta_p_switch = equilibrium_modal_split(rho_d, rho_p)
ax.set_title(r"Equilibrium modal split ($N_p / N = {N_p:.0f}\% $, $N_d / N = {N_d:.0f}\% $)".format(N_p=N_p/N * 100, N_d = N_d/N * 100))

t_d_0 = t_star - (N_d / s) * gamma / (beta + gamma)
t_d_tilde = t_star - (N_d / s) * (beta*gamma) / (alpha * (beta + gamma))
t_d_1 = t_star + (N_d / s) * beta / (beta + gamma)

pos_2 = ax.get_position()
ax.set_position([pos_2.x0, pos_2.y0 - pos_2.height, pos_1.width, pos_1.height * 2])


xticks = {r"$t_d^0$": t_d_0,
         r"$\tilde{t}_d$": t_d_tilde,
         r"$t^*$": t_star,
         r"$t_d^1$": t_d_1
        }

xticklabels = list(xticks.keys())
xtickvalues = list(xticks.values())

yticklabels = ["0", "N"]
ytickvalues = [0, N]

ax.set_xticks(xtickvalues)
ax.set_xticklabels(xticklabels)
ax.set_yticks(ytickvalues)
ax.set_yticklabels(yticklabels)

ax.set_ylim(bottom=0 - N/10, top=N*1.1)
ax.set_xlim(left=t_d_0 - (t_d_1 - t_d_0)/10, right=t_d_1 + (t_d_1 - t_d_0)/10)

x = np.linspace(t_d_0, t_d_1, 1000)
x1 = np.linspace(t_d_0, t_d_tilde, 1000)
x2 = np.linspace(t_d_tilde, t_d_1, 1000)

ax.plot(x, s*(x-t_d_0), color = "black", linestyle="--", label=r"Arrivals (slope = $s$)")
ax.plot(x1, (s + beta * s / (alpha - beta)) * (x1-t_d_0), color="red", label=r"Departures before $\tilde{t}_d$ (slope = $s + \frac{\beta s}{\alpha - \beta}$)")
ax.plot(x2, (s - gamma * s / (alpha + gamma)) * (x2-t_d_tilde) + (t_d_tilde - t_d_0) * (s + beta * s / (alpha - beta)), color = "green", label=r"Departures after $\tilde{t}_d$ (slope = $s - \frac{\gamma s}{\alpha + \gamma}$")

ax.legend(loc="lower right")

ax.set_xlabel("Time (t)")
ax.set_ylabel("Cumulative departures and arrivals")
```

```python
dim = 1000

fig, ax = plt.subplots(tight_layout=True)
ax.set_aspect('equal', adjustable='box')

ax.set_xlabel(r"$\Delta_p^{\text{min}}$")
ax.set_ylabel(r"$\overline{\Delta_d}$")

f_value = 1
E_value = 1
M_value = 3
alpha_value = 2
phi_value = 0.3
beta_value = 1
gamma_value = 4
delta_value = beta_value * gamma_value / (beta_value + gamma_value)

Delta_p_max_minus_min_value = 20
F_function = sp.stats.triang
c=0.7

Delta_p_full_minus_min = F_function.ppf(M_value / (M_value + 1), c=c, loc=0, scale = Delta_p_max_minus_min_value)

# Find the value of SDPmax / SDP frontier
frontier_value = 2 * M_value * phi_value / (2 + M_value + M_value * phi_value)
Delta_p_min_switch = sp.optimize.root_scalar(lambda Delta_p_min: F_function.cdf(f_value + E_value,
                                                                                c=c,
                                                                                loc=Delta_p_min,
                                                                                scale = Delta_p_max_minus_min_value)\
                                                                 - frontier_value,
                                             bracket=[f_value + E_value - Delta_p_full_minus_min, f_value + E_value],
                                             method='bisect')\
                     .root



left_limit = -25
right_limit = 20
bottom_limit = -20
top_limit = 20

ax.set_xlim(left=left_limit, right=right_limit)
ax.set_ylim(bottom=bottom_limit, top=top_limit)

# Zone D and DP frontier
Delta_p_min = np.linspace(left_limit, f_value + E_value, dim)
Delta_d = Delta_p_min - f_value - E_value
ax.plot(Delta_p_min, Delta_d, color="black")

# Zone DP and DPmax frontier
Delta_p_min = np.linspace(left_limit, f_value + E_value - Delta_p_full_minus_min, dim)
Delta_d = Delta_p_min + Delta_p_full_minus_min - f_value - E_value
ax.plot(Delta_p_min, Delta_d, color="black")

# Zone S and SDPmax frontier
Delta_p_min = np.linspace(left_limit, f_value + E_value, dim)
Delta_d = -M_value * (Delta_p_min - f_value - E_value)
ax.plot(Delta_p_min, Delta_d, color="black")

# Zone DPmax and SDPmax frontier
Delta_p_min = np.linspace(left_limit, f_value + E_value - Delta_p_full_minus_min, dim)
Delta_d = -M_value * (Delta_p_min + Delta_p_full_minus_min - f_value - E_value)
ax.plot(Delta_p_min, Delta_d, color="black")

# Zone SD (and S VS D frontier)
Delta_p_min = np.linspace(f_value + E_value, right_limit, dim)
Delta_d = np.zeros(dim)
ax.plot(Delta_p_min, Delta_d, color="black", linestyle=":")


# Zone SDPmax and DP frontier (maybe SDP)
ax.scatter(Delta_p_min_switch, 0, color="black", marker="x", s=100)

Delta_p_min = np.linspace(f_value + E_value - Delta_p_full_minus_min, Delta_p_min_switch, dim)               
Delta_d = np.zeros(dim)
ax.plot(Delta_p_min, Delta_d, color="black", linestyle="-")

Delta_p_min = np.linspace(Delta_p_min_switch, f_value + E_value, dim)               
Delta_d = np.zeros(dim)
ax.plot(Delta_p_min, Delta_d, color="black", linestyle=":")



# Annotation
arrow_length = 5

## Zone D

y = bottom_limit / 2
x = (right_limit + f_value + E_value + y) / 2
ax.text(x, y,
        "D",
        horizontalalignment="center",
        verticalalignment="center",
        fontsize="large",
        fontweight="bold")



## Zone DPmax
yDPmax = -7

y1 = yDPmax
x1 = f_value + E_value + yDPmax - Delta_p_full_minus_min
angle = np.pi / 2
x2 = x1 + arrow_length * np.cos(angle)
y2 = y1 + arrow_length * np.sin(angle)
ax.annotate("", xy=(x1, y1), xytext=(x2, y2), arrowprops=dict(arrowstyle='<-, head_width=0.4, head_length=0.6', color='black', linewidth=3))


y = 0
x = x1
ax.text(x, y,
        "DPmax",
        horizontalalignment="center",
        verticalalignment="center",
        fontsize="large",
        fontweight="bold")

## Zone DP
yDP = -12

y1 = yDP
x1 = f_value + E_value + yDP
angle = np.pi
x2 = x1 + arrow_length * np.cos(angle)
y2 = y1 + arrow_length * np.sin(angle)
ax.annotate("", xy=(x1, y1), xytext=(x2, y2), arrowprops=dict(arrowstyle='<-, head_width=0.4, head_length=0.6', color='black', linewidth=3))


y = yDP
x = f_value + E_value - Delta_p_full_minus_min / 2 + y
ax.text(x, y,
        "DP",
        horizontalalignment="center",
        verticalalignment="center",
        fontsize="large",
        fontweight="bold")


## Zone S
y = top_limit / 2
x = (f_value + E_value + right_limit) / 2
ax.text(x, y,
        "S",
        horizontalalignment="center",
        verticalalignment="center",
        fontsize="large",
        fontweight="bold")

## Zone SD
y = 0
x = (f_value + E_value + right_limit) / 2
ax.annotate("SD",
            xy=(x, 0),
            xytext=(x -3, 5),
            arrowprops=dict(arrowstyle='->', color='black'),
            horizontalalignment="center",
            verticalalignment="center",
            fontsize="large",
            fontweight="bold"
           )


## Zone SDPmax

ySDPmax = 8


y1 = ySDPmax
x1 = f_value + E_value - Delta_p_full_minus_min - y1 / M_value
angle = 0
x2 = x1 + arrow_length * np.cos(angle)
y2 = y1 + arrow_length * np.sin(angle)
ax.annotate("", xy=(x1, y1), xytext=(x2, y2), arrowprops=dict(arrowstyle='<-, head_width=0.4, head_length=0.6', color='black', linewidth=3))

y1 = ySDPmax
x1 = f_value + E_value - y1 / M_value
angle = np.pi
x2 = x1 + arrow_length * np.cos(angle)
y2 = y1 + arrow_length * np.sin(angle)
ax.annotate("", xy=(x1, y1), xytext=(x2, y2), arrowprops=dict(arrowstyle='<-, head_width=0.4, head_length=0.6', color='black', linewidth=3))

y1 = 0
x1 = (f_value + E_value - Delta_p_full_minus_min + Delta_p_min_switch) / 2
angle = np.pi / 2
x2 = x1 + arrow_length * np.cos(angle)
y2 = y1 + arrow_length * np.sin(angle)
ax.annotate("", xy=(x1, y1), xytext=(x2, y2), arrowprops=dict(arrowstyle='<-, head_width=0.4, head_length=0.6', color='black', linewidth=3))

y = ySDPmax + 3
x1 = f_value + E_value - y/M_value - Delta_p_full_minus_min
x2 = f_value + E_value - y/M_value
ax.text((x2 + x1)/2, y,
        "SDPmax",
        horizontalalignment="center",
        verticalalignment="center",
        fontsize="large",
        fontweight="bold")





x = (Delta_p_min_switch + f_value + E_value) / 2
ax.annotate("SDP",
            xy=(x, 0),
            xytext=(x -7, -7),
            arrowprops=dict(arrowstyle='->', color='black'),
            horizontalalignment="center",
            verticalalignment="center",
            fontsize="large",
            fontweight="bold"
           )



# Measures
y = 15
x1 = f_value + E_value - y/M_value - Delta_p_full_minus_min
x2 = f_value + E_value - y/M_value
ax.annotate('', xy=(x1, y), xytext=(x2, y),
            arrowprops=dict(arrowstyle='<->', color='black'))
ax.text((x2 + x1)/2, y + (top_limit - bottom_limit) * 0.02, r"$\Delta_p^{\text{full}} - \Delta_p^{\text{min}}$", horizontalalignment="center")

y = 13
x = f_value + E_value - y/M_value
ax.annotate(r'$slope = -M$', xy=(x, y), xytext=(x+10, y + 4),
            arrowprops=dict(arrowstyle="->", linestyle="--", color='black'),
            fontsize="small"
           )

y = -7
x = f_value + E_value + y
ax.annotate(r'$slope = 1$', xy=(x, y), xytext=(x+4, y - 10),
            arrowprops=dict(arrowstyle="->", linestyle="--", color='black'),
            fontsize="small"
           )

xtickvalues = [f_value + E_value - Delta_p_full_minus_min, f_value + E_value]
xticklabels = [r"$f + E - (\Delta_p^{\text{full}} - \Delta_p^{\text{min}})$", r"$f + E$"]
ytickvalues = [0]

ax.set_xticks(xtickvalues)
ax.set_xticklabels(xticklabels)
ax.set_yticks(ytickvalues)
```
