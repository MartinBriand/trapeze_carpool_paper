## Infinite demand of passengers with null benefit, modal choice reduced between sharing seats on carpool lane or solo-driving on general purpose lane

### Mémo


### Setting

We have $N$ drivers choosing either to drive solo ($N_s$ solo drivers), or to share their seats while using the reserved lane ($N_d$ carpool drivers). Since there is an infinite demand from passengers, the carpooling drivers always incur the inconvenience cost $\Delta$.

Therefore, the cost incurred by solo-drivers is

$$
c_s(t) = \alpha q_g(t) + \max [\beta(t^* - t - q_g(t)), \gamma(t+ q_g(t) - t^*)] + f
$$

And the cost incurred by carpool drivers is 

$$
c_d(t) = \alpha q_c(t) + \max [\beta(t^* - t - q_c(t)), \gamma(t+ q_c(t) - t^*)] + f + \Delta + \rho_d(t)
$$

Therefore, the social-cost to minimize is 

$$
SC = \underbrace{\int_{time} c_s(t) r_s(t)dt}_{=SC_s} + \underbrace{\int_{time} c_d(t) r_d(t)dt - \Pi}_{=SC_d}\\
$$

where $\Pi$ is the profit of the carpooling platform operating the reserved lane, namely:

$$
\Pi = \rho_d(t)r_d(t)
$$

### Solving for a fixed $\varphi$

Suppose we have a fixed $\varphi \in [0, 1]$. Suppose that the $N$ drivers split between the two modes like this: $N_d \in [0, N]$, $N_s \in [0, N]$ and $N_d + N_s = N$. It is clear that it is impossible to have $N_d > 0$ if $\varphi = 0$ as well as it is impossible to have $N_s >0$ if $\varphi = 1$.

Also, suppose that $\varphi < 1$, then, for the $N_s$ drivers on the general purpose lane, since they do not incur any kind of toll, it is straightforward to see that the solution for them is a very classical bottleneck model:

$$
\left\{
\begin{array}{ll}
\dot{q}_g(t) = \frac{\beta}{\alpha - \beta} & \text{if } t+q_g(t) \leq t^*\\
\dot{q}_g(t) = -\frac{\gamma}{\alpha + \gamma} & \text{if } t+q_g(t) \geq t^*
\end{array}
\right.\\

\left\{
    \begin{array}{ll}
        r_g (t) = r_s(t) =s + \frac{\beta s}{\alpha - \beta} & \text{if } t+q_g(t) \leq t^*\\
        r_g (t) = r_s(t) =s - \frac{\gamma s}{\alpha + \gamma} & \text{if } t+q_g(t) \geq t^* 
    \end{array}
\right.\\

\left\{
    \begin{array}{l}
        t_s^0 = t^* - \frac{\gamma}{\beta + \gamma} \cdot \frac{N_s}{(1-\varphi)s}\\
        \tilde{t}_s = t^* - \frac{\beta \gamma}{\alpha(\beta + \gamma)}\cdot \frac{N_s}{(1-\varphi)s}\\
        t_s^1 = t^* + \frac{\beta}{\beta + \gamma} \cdot \frac{N_s}{(1-\varphi)s}
    \end{array}
\right.\\

\left\{
    \begin{array}{l}
        c_s(t) = c_s = \delta \frac{N_s}{(1-\varphi)s} + f\\
        SC_s = \delta \frac{N_s^2}{(1-\varphi)s} + N_sf
    \end{array}
\right. \text{where } \delta = \frac{\beta \gamma}{\beta + \gamma}
$$

#### Minimizing Social Cost
##### The Maths
###### Properties to simplify

Now, suppose that we still have our fixed $\varphi \in [0,1]$, but suppose this time that $\varphi > 0$. Then, for the $N_d$ drivers on the reserved lane, in a very classical bottleneck model way, one can minimize $SC_d$ my setting the following fare (or toll): $\rho_d(t) = \rho_d^0 + \rho_d^*$ where $\rho_d^0$ is an arbitrary constant (positive or negative), and

$$
\left\{
    \begin{array}{ll}
        \rho_d^*(t) = 0 & \text{if } t<t_d^0\\
        \rho_d^*(t) = \delta \frac{N}{s} - \beta (t^* - t) & \text{if } t\in [t_d^0, t^*]\\
        \rho_d^*(t) = \delta \frac{N}{s} - \gamma (t-t^*) & \text{if } t \in [t^*, t_d^1]\\
        \rho_d^*(t) = 0 & \text{if } t>t_d^1
\end{array}
\right.
$$

where

$$
\left\{
    \begin{array}{l}
        t_d^0 = t^* - \frac{\gamma}{\beta + \gamma} \cdot \frac{N_d}{\varphi s}\\
        t_d^1 = t^* + \frac{\beta}{\beta + \gamma} \cdot \frac{N_d}{\varphi s}
    \end{array}
\right.
$$


We hence obtain the minimized Social Cost and incurred cost for carpool drivers:

$$
\left\{
    \begin{array}{l}
        c_d(t) = c_d = \delta \frac{N_d}{\varphi s} + f + \Delta + \rho_d^0\\
        SC_d = \delta \frac{N_d^2}{2 \varphi s} + N_d f + N_d \Delta
    \end{array}
\right.
$$

We then easily get the following property

**Property:**

For any fare / toll $\rho_d ' (t)$ that minimizes the total Social Cost of the $N = N_s + N_d$ users. And let $(N_s, N_d)$ (with $N_s + N_d = N$) be the modal split obtain with this fare / toll.

Then, there exist a $\rho_d^0 \in \R$ such that the fare - toll $\rho_d(t) = \rho_d^0 + \rho_d^*(t)$ (like above) do just as good with the same modal split.

***Proof:***

Suppose $\rho_d ' (t)$ results in mode splid $(N_d, N_s)$ (with $N_d + N_s * N$). The Wardrope equilibrium first principle requires that $c_s(t) = c_d(t) = c$. Let $\rho_d^0 = c - \delta \frac{N}{\varphi s}$. Switching from $\rho_d '(t)$ to $\rho_d(t) = \rho_d^0 + \rho_d^*(t)$, we easily see that the costs incurred by drivers remain unchanged. Therefore, the modal split should not change either. Also, the Social Cost and the general lane remains unchanged, and the Social Cost on the reserved lane in now optimized. Therefore, the total social can only be lower, but since it was already optimized, it remains unchanged. &square;

With such a fare, social and individual costs can now be summarized like below:

$$
\left\{
    \begin{array}{l}
        SC = \frac{\delta}{s}\left( \frac{N_d^2}{2 \varphi} + \frac{N_s^2}{1 - \varphi}\right) + N f + N_d \Delta\\
        c_s(t) = c_s = \delta \frac{N_s}{(1-\varphi)s} + f\\
        c_d(t) = c_d = \delta \frac{N_d}{\varphi s} + f + \Delta + \rho_d^0
    \end{array}
\right.
$$

###### Solving the best modal split

We can now obtain the following corollary:

**Corollary**:

* If $\varphi \in [0, 1] \setminus \{0, 1\}$, finding the socially optimum fare / toll is equivalent to finding the optimal $\rho_d^0$ that minimizes the Social Cost just above, which (knowing that any modal split could be obtained from $\rho_d^0$ values, is equivalent to finding the optimal modal split that minimizes the Social Cost above).
* If $\varphi = 0$, $N_d = 0$, $N_s =N$, $c_d$ is undefined, $c_s = \delta \frac{N}{(1-\varphi)s} + f$ and $SC = \delta \frac{N^2}{(1 - \varphi)s} + N f$
* If $\varphi = 1$, $N_s = 0$, $N_d =N$, $c_s$ is undefined, $c_d = \delta \frac{N}{\varphi s} + f + \Delta + \rho_d^0$ and $SC = \delta \frac{N^2}{2 \varphi s} + N f + N \Delta$

We therefore only use the kind of fare / toll presented above for what follows and minimize Social Cost by finding the optimal split given the Social Cost function above.

Since the cases where $\varphi \in \{0, 1\}$ is already solved, we can now suppose $\varphi \in [0, 1] \setminus \{0, 1\}$.

We will solve the following Lagrangian:

$$
\mathcal{L} = \frac{\delta}{s}\left( \frac{N_d^2}{2\varphi} + \frac{N_s^2}{1-\varphi} \right) + N f + N_d \Delta\\
-\lambda (N-N_s - N_d)\\
-\mu _1 (N_s - \alpha _1^2)
-\mu _2 (N_d - \alpha _2^2)
$$

Where $\lambda$ is the Lagrange multiplier for an equality constraint, and $\mu _1$ and $\mu _2$ are Lagrange multipliers for the two equality constraints, and $\alpha _1$ and $\alpha _2$ are slack variables for the inequality constraints.

Differentiating the Lagrangian, we get:

$$
\left\{
    \begin{array}{lr}
        \frac{\partial \mathcal{L}}{\partial N_d} = \frac{\delta}{s} \cdot \frac{N_d}{\varphi} + \Delta + \lambda - \mu _2 = 0 & (1)\\
        \frac{\partial \mathcal{L}}{\partial N_s} = \frac{\delta}{s} \cdot \frac{2N_s}{1-\varphi} + \lambda - \mu _1 = 0 & (2)\\
        \frac{\partial \mathcal{L}}{\partial \lambda} = N_s + N_d - N = 0 & (3)\\
        \frac{\partial \mathcal{L}}{\partial \mu _1} = \alpha _1^2 - N_s = 0 & (4)\\
        \frac{\partial \mathcal{L}}{\partial \mu _2} = \alpha _2^2 - N_d = 0 & (5)
    \end{array}
\right.
$$

We then have four cases:

1. $\mu _1 > 0$, $N_s = 0$, $\mu _2 > 0$ and $N_d = 0$.
1. $\mu _1 = 0$, $N_s \geq 0$, $\mu _2 > 0$ and $N_d = 0$.
1. $\mu _1 > 0$, $N_s = 0$, $\mu _2 = 0$ and $N_d \geq 0$.
1. $\mu _1 = 0$, $N_s \geq 0$, $\mu _2 = 0$ and $N_d \geq 0$.

**Case 1** is clearly impossible.

**Case 2:** As said, we have $N_d = 0$. From (3), we get $N_s = N$. From (2), we get $\lambda = -\frac{2\delta N}{(1-\varphi)s}$ (which is logically negative since reducing the number of participants would otherwise be an extremely effective way to reduce costs). Finally, from (1) and (2), we get $\mu _2 = \Delta - \frac{2\delta N}{s(1-\varphi)}$.

And then, the condition $\mu _2 > 0$ is equivalent to $\Delta > \frac{2\delta N}{s(1-\varphi)}$.

**Case 3:** As said, we have $N_s = 0$. From (3), we get $N_d = N$. From (1), we get $\lambda = -\Delta - \frac{\delta N}{\varphi s}$. From (2), we get $\mu _1 = -\Delta - \frac{\delta N}{\varphi s}$

And then, the condition $\mu _1 > 0$ is equivalent to $-\Delta > \frac{\delta N}{\varphi s}$.

> We see that this time, $\lambda >0$. This means that carpooling is actually so attractive that people would see a carpool commute beneficial despite the fact that they lose time in transportation (for the pleasure of meeting someone is extremely high). Actually, if we were to have calculated our Lagrangian the proper way with $(N_d + N_s)f$ instead of $Nf$, our lambda would have been lowered by $f$, and therefore, if $f$ is high enough, more commute would not be beneficial anymore.

**Case 4:** With (1), (2) and (3), we get
$$
\left\{
    \begin{array}{l}
        N_d = \frac{\varphi}{1+\varphi}(2N - \frac{\Delta s}{\delta}(1-\varphi))\\
        N_s = \frac{1- \varphi}{1+\varphi}(N + \frac{\Delta s}{\delta}\varphi)\\
        \lambda = -\frac{2}{1 + \varphi} \left( \frac{\delta N}{s} + \Delta \varphi \right)
    \end{array}
\right.
$$

Then, the two conditions $N_d \geq 0$ and $N_s \geq 0$ give $-\frac{\delta N}{\varphi s} \leq \Delta \leq \frac{2\delta N}{(1-\varphi)s}$

> It is now clear that $\lambda \leq 0$, so reducing the traffic could only be beneficial in terms of social cost, while this is constrained by the inelastic demand $N$.

> There is an intuitive interpretation in the inequality constraint for the interior solutions:
> * If everyone is on the reserved lane, then the marginal time social cost (with optimal fare) is $\frac{\delta N}{\varphi s}$. Then, if a marginal user were to switch lane. She would save the cost $\Delta$ and reduce the time social cost by the marginal time social cost *supra*, which is non-beneficial only if $-\Delta > \frac{\delta N}{\varphi s}$
> * If everyone is on the GP lane, then the marginal time social cost (without optimal fare) is $\frac{2 \delta N}{(1-\varphi) s}$. Then, if a marginal user were to switch lane. She would incure the cost $\Delta$ and reduce the time social cost by the marginal time social cost *supra*, which is non-beneficial only if $\Delta < \frac{2\delta N}{(1-\varphi) s}$

###### Revenues of the operator
The question now is. What is the value of $\rho_d^0$ that allows for those optimal splits? And what is the potential revenues of the platform.

Using $c_s = c_d \Leftrightarrow \frac{\delta N_s}{(1-\varphi)s} + f = \frac{\delta N_d}{\varphi s} + f + \Delta + \rho_d^0$ (for case 4, for case 2, replace "$=$" by "$\leq$", and by "$\geq$" for case 3), and knowing that the average revenue per carpooling user from $\rho_d^*$ is $\frac{\delta N_d}{\varphi s}, we get:

* For case 2 (nobody carpools): $\rho_d^0 \geq -\Delta + \frac{\delta N}{(1-\varphi)s}$ and $R = 0$
* For case 3 (everybody carpools): $\rho_d^0 \leq -\Delta - \frac{\delta N}{\varphi s}$ (toll can be positive since the RHS is positive according to conditions) and $R \leq N\left( -\Delta - \frac{\delta N}{2\varphi s} \right)$ (which could be positive)
* For case 4 (there is a split): $\rho_d^0 = \frac{1}{1+\varphi}\left( -\Delta \varphi - \frac{\delta N}{s} \right)$ and $R = \frac{N_d}{2(1+\varphi)} \times (3-\varphi) \times (-\Delta)$ (which is of the sign a $\Delta$ meaning that if $\Delta$ is positive, the service should be subsided, and if $\Delta$ is negative, the service can raise revenues).

##### Computational checks
###### For the modal split

Here we do two graphs, one with the maths, and one with numerical calculations to make sure we have close results.

```python
import numpy as np

import matplotlib.pyplot as plt
from matplotlib import cm
from matplotlib.colors import LightSource

import pandas as pd
```

```python
alpha = 2.
beta = 1.
gamma = 4.

delta = beta * gamma / (beta + gamma)

N = 10000.
s = 4000.

phis = np.linspace(0, 1, 101)

carpoolcost_total_cost_ratios = np.linspace(-10, 10, 101)
```

```python
df = pd.DataFrame({"phi": phis})\
    .merge(pd.DataFrame({"carpoolcost_total_cost_ratio": carpoolcost_total_cost_ratios}),
           how="cross")

df["big_delta"] = df.carpoolcost_total_cost_ratio * delta * N / s

df["set_to_one"] = False
df.loc[df.phi == 1, "set_to_one"] = True
df.loc[(df.phi != 0)\
       &(df.phi != 1)\
       , "set_to_one"] = df.big_delta < -delta * N / (df.phi * s)

df["set_to_zero"] = False
df.loc[df.phi == 0, "set_to_zero"] = True
df.loc[(df.phi != 0)\
       &(df.phi != 1)\
       , "set_to_zero"] = df.big_delta > 2 *delta * N / ((1-df.phi) * s)

df["calculate"] = (~df.set_to_one) & (~df.set_to_zero)

df["Nd_over_N_ratio"] = -1.
df.loc[df.set_to_zero, "Nd_over_N_ratio"] = 0
df.loc[df.set_to_one, "Nd_over_N_ratio"] = 1
df.loc[df.calculate, "Nd_over_N_ratio"] = (2.*N - df.loc[df.calculate, "big_delta"]*(1.*s)*(1.-df.loc[df.calculate, "phi"])/delta)\
       * df.loc[df.calculate, "phi"] / (1. + df.loc[df.calculate, "phi"]) / N

```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 16))


x = df.phi.to_numpy().reshape((101, 101))
y = df.carpoolcost_total_cost_ratio.to_numpy().reshape((101, 101))
z = df.Nd_over_N_ratio.to_numpy().reshape((101, 101))

# ax1
ax1.set_xlabel(r"$\varphi$")
ax1.set_ylabel(r"$\frac{N_d}{N}$")

index1 = (slice(None), slice(35, -34, 3))
x1 = x[index1]
y1 = y[index1][0]
z1 = z[index1]

ax1.plot(x1, z1, label=y1)

ax1.legend(title=r"Value of $\frac{\Delta}{\frac{\delta N}{s}}$")
ax1.grid(True)

# ax2
ax2.set_xlabel(r"$\frac{\Delta}{\frac{\delta N}{s}}$")
ax2.set_ylabel(r"$\frac{N_d}{N}$")

index2 = (slice(None, None, 10), slice(None, None, None))
x2 = x[index2][:, 0]
y2 = y[index2]
z2 = z[index2]

ax2.plot(np.transpose(y2), np.transpose(z2), label=x2)

ax2.legend(title=r"Value of $\varphi$")
ax2.grid(True)
```

```python
df2 = df.merge(pd.DataFrame({"Nd_test": np.linspace(0, N, 101)}),
              how="cross")

df2["SC_test"] = -1000000.

df2 = df2.loc[(df2.phi > 0.) | (df2.Nd_test == 0. )].copy()
df2.loc[df2.phi == 0, "SC_test"] = delta * N * N / ((1-df2.phi)*s) # + N*f

df2 = df2.loc[(df2.phi < 1.) | (df2.Nd_test == N)].copy()
df2.loc[df2.phi == 1, "SC_test"] = delta * N * N / (2*df2.phi*s) + N*df2.big_delta # + N*f

df2.loc[(df2.phi > 0.) & (df2.phi < 1.), "SC_test"] = (delta / s) * (df2.Nd_test * df2.Nd_test / (2*df2.phi) + (N-df2.Nd_test) * (N-df2.Nd_test) / (1-df2.phi)) + df2.Nd_test*df2.big_delta # +N*f

df2_argmin = df2.groupby(["phi", "carpoolcost_total_cost_ratio"])[["SC_test"]]\
    .idxmin()\
    .rename(columns={"SC_test": "argmax_index"})

df2_argmin["Nd_over_N_ratio_best_compute"] = (df2.loc[df2_argmin.argmax_index, "Nd_test"] / N).to_numpy()

df2_argmin.reset_index(inplace=True)
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 16))


x = df2_argmin.phi.to_numpy().reshape((101, 101))
y = df2_argmin.carpoolcost_total_cost_ratio.to_numpy().reshape((101, 101))
z = df2_argmin.Nd_over_N_ratio_best_compute.to_numpy().reshape((101, 101))

# ax1
ax1.set_xlabel(r"$\varphi$")
ax1.set_ylabel(r"$\frac{N_d}{N}$")

index1 = (slice(None), slice(35, -34, 3))
x1 = x[index1]
y1 = y[index1][0]
z1 = z[index1]

ax1.plot(x1, z1, label=y1)

ax1.legend(title=r"Value of $\frac{\Delta}{\frac{\delta N}{s}}$")
ax1.grid(True)

# ax2
ax2.set_xlabel(r"$\frac{\Delta}{\frac{\delta N}{s}}$")
ax2.set_ylabel(r"$\frac{N_d}{N}$")

index2 = (slice(None, None, 10), slice(None, None, None))
x2 = x[index2][:, 0]
y2 = y[index2]
z2 = z[index2]

ax2.plot(np.transpose(y2), np.transpose(z2), label=x2)

ax2.legend(title=r"Value of $\varphi$")
ax2.grid(True)
```

The two graphs are extremely close so we can consider the formulas as correct.

###### For the values of $\rho_d^0$ and revenues

We check that $\rho_d^0$ has the correct equality or inequality constraint. And same for revenues.

```python
df3 = df.loc[(0. < df.phi) & (df.phi < 1.)].copy()

df3["should_be_bigger"] = df3.big_delta > 2 * delta * N / (s* (1- df3.phi))

df3["should_be_lower"] = -df3.big_delta > delta * N / (s* df3.phi)

df3["should_be_equal"] = (~df3.should_be_lower) & (~df3.should_be_bigger)

df3["rhod0"] = 0.
df3.loc[df3.should_be_bigger, "rhod0"] = -df3.big_delta + delta * N / (s* (1- df3.phi))
df3.loc[df3.should_be_lower, "rhod0"] = -df3.big_delta - delta * N / (s* df3.phi)
df3.loc[df3.should_be_equal, "rhod0"] = (1/(1+ df3.phi)) * (-df3.big_delta * df3.phi - delta * N / s)

df3["cs"] = delta * N * (1-df3.Nd_over_N_ratio) / ((1 - df3.phi) * s)
df3["cd"] = delta * N * df3.Nd_over_N_ratio / (df.phi * s) + df3.big_delta + df3.rhod0

eps = 0.001
df3["is_correct_rho"] = False
df3.loc[df3.should_be_bigger, "is_correct_rho"] = df3.cd + eps > df3.cs
df3.loc[df3.should_be_lower, "is_correct_rho"] = df3.cd - eps < df3.cs
df3.loc[df3.should_be_equal, "is_correct_rho"] = (df3.cd -df3.cs).abs() < eps

print("Rho test passed" if df3.is_correct_rho.all() else "rho test did not pass")


df3["calculated_revenue"] = (df3.rhod0 + delta * df3.Nd_over_N_ratio * N / (2 * df3.phi * s))* df3.Nd_over_N_ratio * N

df3["math_revenue"] = 0.
df3.loc[df3.should_be_bigger, "math_revenue"] = 0.
df3.loc[df3.should_be_lower, "math_revenue"] = N * (-df3.big_delta - delta * N / (2*df.phi * s))
df3.loc[df3.should_be_equal, "math_revenue"] = df3.Nd_over_N_ratio * N / (2*(1+ df3.phi)) * (1 + df3.phi) * (-df3.big_delta)

df3["is_correct_revenue"] = False
df3.loc[df3.should_be_bigger, "is_correct_revenue"] = df3.calculated_revenue + eps > df3.math_revenue
df3.loc[df3.should_be_lower, "is_correct_revenue"] = df3.calculated_revenue - eps < df3.math_revenue
df3.loc[df3.should_be_equal, "is_correct_revenue"] = (df3.calculated_revenue -df3.math_revenue).abs() < eps

print("Revenue test passed" if df3.is_correct_revenue.all() else "revenue test did not pass")
```
