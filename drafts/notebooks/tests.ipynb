```python
import numpy as np
from scipy.stats import anglit, alpha, beta, betaprime, norm, lognorm, logistic
import matplotlib.pyplot as plt
```

```python
# norm

delta_d = 1
fuel = 0
rho_d = 0
rho_p = 2


delta_p = np.linspace (-3, 1, 1000)
F = norm.cdf
y_d = rho_d + fuel + delta_d * F(delta_p) / (1-F(delta_p))
y_p = delta_p + rho_p

fig, ax = plt.subplots(1, 1)
ax.plot(delta_p, y_p, label=r"$y_p$")
ax.plot(delta_p, y_d, label=r"$y_d$")

ax.legend()
```

```python
# lognorm

delta_d = 1
fuel = 0
rho_d = 0
rho_p = 2


delta_p = np.linspace (-3, 5, 1000)
F = lognorm(s=1).cdf
y_d = rho_d + fuel + delta_d * F(delta_p) / (1-F(delta_p))
y_p = delta_p + rho_p

fig, ax = plt.subplots(1, 1)
ax.plot(delta_p, y_p, label=r"$y_p$")
ax.plot(delta_p, y_d, label=r"$y_d$")

ax.legend()
```

```python
# beta

delta_d = 1
fuel = 0
rho_d = 0
rho_p = 2


delta_p = np.linspace (-3, 1, 1000)
F = beta(a=5, b=1).cdf
y_d = rho_d + fuel + delta_d * F(delta_p) / (1-F(delta_p))
y_p = delta_p + rho_p

fig, ax = plt.subplots(1, 1)
ax.plot(delta_p, y_p, label=r"$y_p$")
ax.plot(delta_p, y_d, label=r"$y_d$")

ax.set_ybound((-5, 5))

ax.legend()
```
